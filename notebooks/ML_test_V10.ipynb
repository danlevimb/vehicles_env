```python
# Ejemplo de Métricas de clasificación: Estudio de probabilidades
# Importar librerías 

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
```

```python
# Ruta del dataset
file_path = "C:/Git/vehicles_env/datasets/Electrical_Grid_Stability_us.csv"

# Cargar dataframe
electrical_grid = pd.read_csv(file_path, sep=';')

print('Tamaño del dataframe:', electrical_grid.shape)
#print(electrical_grid.head())

print('La relación entre las clases: \n', electrical_grid['stability'].value_counts())
```

```python
# Divide los datos en características (la matriz X) y una variable objetivo (y)
X = electrical_grid.drop('stability', axis = 1)
y = electrical_grid['stability']

# Dividir los datos de entrenamiento (train) y datos de prueba (test)
# Especificar que el tamaño de los datos para prueba será el 20% de los registros
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Especificar el tipo de modelo
model = LogisticRegression()

# Entrenar el modelo con los conjuntos de entrenamiento (train)
model.fit(X_train, y_train)

# Estima la probabilidad de que un objeto entre en la segunda clase
probabilities = model.predict_proba(X_test)[:,1]

# Calcular la métrica
# Receiver Operating Characteristic (ROC) & Area Under Curve (AUC)
roc_auc = roc_auc_score(y_test, probabilities)

# muestra todas las métricas estudiadas para la predicción resultante
print('ROC_AUC: {:.2f}'.format(roc_auc))
```

¡Es correcto!

La métrica roc_auc se ve muy bien, ¡casi 0.9! El modelo parece predecir la estabilidad bastante bien.
