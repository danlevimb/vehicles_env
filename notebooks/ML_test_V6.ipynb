```python
# Ejemplo de modelo linear Lasso
# Importar librerías 

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
```

```python
# Ruta del dataset
file_path = "C:/Git/vehicles_env/datasets/dataset_facebook_cosmetics_us.csv"

# Cargar dataframe
fb = pd.read_csv(file_path, sep=';')
```

```python
# divide los datos en características (la matriz X) y una variable objetivo (y)
X = fb.drop('Total Interactions', axis = 1)
y = fb['Total Interactions']

# Dividir los datos de entrenamiento y datos de prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Crear la instancia del objeto estandarizador
scaler = StandardScaler()

# Entrenar el scaler en el conjunto de entrenamiento usando el método fit
scaler.fit(X_train)

# Estandarizar el conjunto de entrenamiento
X_train_st = scaler.transform(X_train)

# Estandarizar el conjunto de prueba
X_test_st = scaler.transform(X_test)

# Definir el modelo 
model = Lasso()

# Entrenar el modelo con los datos de entrenamiento (Estandarizados)
model.fit(X_train_st, y_train)

# Crear predicción para los datos de validación (Estandarizados)
predictions = model.predict(X_test_st)

# Crear dataframe con características y sus pesos
features = pd.DataFrame({'feature':X_train.columns, 'coeff':model.coef_})

# Crear la columna con los pesos (absolutos)
features['coeff_abs'] = [abs(i) for i in features['coeff']]

# Imprimir las características y su peso
print(features.sort_values('coeff_abs', ascending=False))
```

¡Es correcto!

El peso con el valor absoluto más alto corresponde al ciclo de vida promedio de los/las usuarios/as activos/as. La explicación probablemente sea que cuanto más tiempo permanezcan contigo los/las usuarios/as activos/as, más fieles serán al contenido que ofreces.
