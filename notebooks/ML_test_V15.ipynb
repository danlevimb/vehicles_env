```python
# Ejemplo de Cálculo del valor de la silueta para el agrupamiento
# Importar librerías 

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# Leer datos
file_path = "C:/Git/vehicles_env/datasets/tripadvisor_review_case_us.csv"
travel = pd.read_csv(file_path) # Cargar dataframe
print(travel.shape) # Mostrar el tamaño del dataframe
```

```python
# estandariza los datos
sc = StandardScaler()
X_sc = sc.fit_transform(travel)

# define el modelo k_means con 3 clústeres
km = KMeans(n_clusters = 3)
# predice los clústeres para las observaciones (el algoritmo les asigna un número de 0 a 4)
labels = km.fit_predict(X_sc)

# almacena etiquetas de clúster en el campo de nuestro conjunto de datos
travel['cluster_km'] = labels

# calcula el valor de la silueta para el agrupamiento
print('silhouette_score: {:.2f}'.format(silhouette_score(X_sc, labels)))
```

¡Es correcto!

El resultado resultó ser relativamente bajo. Pero recuerda que en los gráficos pareados no había grupos claramente distinguibles, por lo que no es una sorpresa.

A diferencia de los ejemplos de aprendizaje automático, en los negocios es poco probable que encuentres un proceso en el que las observaciones (objetos) tengan una estructura y una división claras. Por ese motivo, los resultados del agrupamiento a menudo se discuten con expertos y expertas y se evalúan en el curso del análisis.

Este valor del **silhouette\_score = 0.13** te dice mucho sobre la **calidad del agrupamiento** que lograste con K-Means.

---

### 🔹 Qué mide la silueta

* El *silhouette score* varía entre **-1 y 1**:

  * **Cerca de 1** → los puntos están bien agrupados: cada cliente se parece mucho a los de su grupo y poco a los de otros grupos.
  * **Cerca de 0** → los clústeres se mezclan bastante, los puntos no tienen separación clara.
  * **Negativo** → los puntos están más cerca de otros clústeres que del suyo → mala agrupación.

---

### 🔹 Interpretación de tu resultado (0.13)

* **0.13 está apenas arriba de 0**, lo que significa que **los clústeres no están bien definidos**.
* En términos prácticos:

  * Los grupos que encontraste **existen, pero no son muy distintos entre sí**.
  * Hay bastante traslape entre clústeres: algunos clientes podrían pertenecer a otro grupo sin gran diferencia.

---

### 🔹 Qué le puedes decir al cliente

En palabras simples, puedes decir:

Esto significa que los clientes no están perfectamente divididos en segmentos muy distintos; más bien tienen comportamientos que se mezclan. Para mejorar esto, podríamos probar con otro número de clústeres o con otras variables que ayuden a diferenciar mejor los perfiles.”
