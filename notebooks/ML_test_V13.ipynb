```python
# Ejemplo de Bosques aleatorios / Gradientes
# Importar librerías 

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import  accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
# define la función que podrá generar nuestras métricas
def print_all_metrics(y_true, y_pred, y_proba, title = 'Métricas de clasificación'):
    print(title)
    print('\tAccuracy: {:.2f}'.format(accuracy_score(y_true, y_pred)))
    print('\tPrecision: {:.2f}'.format(precision_score(y_true, y_pred)))
    print('\tRecall: {:.2f}'.format(recall_score(y_true, y_pred)))
    print('\tF1: {:.2f}'.format(f1_score(y_true, y_pred)))
    print('\tROC_AUC: {:.2f}'.format(roc_auc_score(y_true, y_proba)))
    print('\n')
```

```python
# Ruta del dataset
file_path = "C:/Git/vehicles_env/datasets/Electrical_Grid_Stability_us.csv"

# Cargar dataframe
electrical_grid = pd.read_csv(file_path, sep=';')

print('Tamaño del dataframe:', electrical_grid.shape)
#print(electrical_grid.head())

print('La relación entre las clases: \n', electrical_grid['stability'].value_counts())
```

```python
# Divide los datos en características (la matriz X) y una variable objetivo (y)
X = electrical_grid.drop('stability', axis = 1)
y = electrical_grid['stability']

# Dividir los datos de entrenamiento (train) y datos de prueba (test)
# Especificar que el tamaño de los datos para prueba será el 20% de los registros
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

#------------------------ ESTANDARIZAR DATOS
scaler = StandardScaler() # Definir el modelo 
scaler.fit(X_train) # Entrenar el modelo con los datos de entrenamiento
X_train_st = scaler.transform(X_train) # Obtener el conjunto de entranamiento estandarizado
X_test_st = scaler.transform(X_test) # Obtener el conjunto de prueba estandarizado

#------------------------ MODELO LINEAL (REGRESIóN LOGíSTICA)
lr_model = LogisticRegression(random_state=0) # Define el arlgoritmo para la regresión logística
lr_model.fit(X_train_st, y_train) # Entrenar el modelo con la matriz de entrenamiento estandarizada y los datos de entrenamiento
lr_predictions = lr_model.predict(X_test_st) # Hacer predicciones
lr_probabilities = lr_model.predict_proba(X_test_st)[:,1] # Hacer probabilidades
print_all_metrics(y_test, lr_predictions, lr_probabilities, title='Métricas de regresión logística')

#------------------------ MODELO ARBOL DE DECISIóN
tree_model = DecisionTreeClassifier() # Definir el modelo
tree_model.fit(X_train_st, y_train) # Entrenar el modelo
tree_predictions = tree_model.predict(X_test_st) # Hacer predicciones
tree_probabilities = tree_model.predict_proba(X_test_st)[:,1] # Hacer probabilidades
print_all_metrics(y_test, tree_predictions, tree_probabilities, title='Métricas del árbol de decisión')

#------------------------ MODELO BóSQUE ALEATORIO
rf_model = RandomForestClassifier(random_state=0) # Definir el modelo
rf_model.fit(X_train_st, y_train) # Entrenar al modelo
rf_predictions = rf_model.predict(X_test_st) # Hacer predicciones 
rf_probabilities = rf_model.predict_proba(X_test_st)[:,1] # Hacer probabilidades 
print_all_metrics(y_test, rf_predictions, rf_probabilities, title='Métricas de Random Forest')

#------------------------ MODELO DE POTENCIACIóN DE GRADIENTE (GRADIENT BOOSTER)
gb_model = GradientBoostingClassifier(random_state=0) # Definir el modelo
gb_model.fit(X_train_st, y_train) # Entrenar al modelo
gb_predictions = gb_model.predict(X_test_st) # Hacer predicciones 
gb_probabilities = gb_model.predict_proba(X_test_st)[:,1] # Hacer probabilidades 
print_all_metrics(y_test, gb_predictions, gb_probabilities, title='Métricas de Potenciación de gradiente')
```
