```python
# Este es el primewr ejemplo de como entrenar un modelo de Machine Learning.
# Importar librerías 

import pandas as pd
from sklearn.ensemble import RandomForestRegressor
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Ruta del dataset
file_path = "C:/Git/vehicles_env/datasets/dataset_facebook_cosmetics_us.csv"

# Cargar dataframe
fb = pd.read_csv(file_path, sep=';')
```

```python
# Crear el dataframe con las características
X = fb.drop(columns='Total Interactions')

# Crear el dataframe con los valores objetivo
y = fb['Total Interactions']
```

```python
# Definir el algoritmo para el modelo
model = RandomForestRegressor()

# Entrenar el modelo
model.fit(X,y)

# Elaborar predicción
predictions = model.predict(X)

sns.scatterplot(x=y, y=predictions, s=15, alpha=0.6)
plt.title('Pronóstico/hecho')
plt.ylabel('Pronóstico')
plt.xlabel('Hecho')
plt.show()
```
