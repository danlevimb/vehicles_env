```python
# Ejemplo de Umbrales y balance de clases
# Importar librerías 

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import  accuracy_score, precision_score, recall_score, f1_score
```

```python
# Ruta del dataset
file_path = "C:/Git/vehicles_env/datasets/Electrical_Grid_Stability_us.csv"

# Cargar dataframe
electrical_grid = pd.read_csv(file_path, sep=';')

print('Tamaño del dataframe:', electrical_grid.shape)
#print(electrical_grid.head())

print('La relación entre las clases: \n', electrical_grid['stability'].value_counts())
```

```python
# Divide los datos en características (la matriz X) y una variable objetivo (y)
X = electrical_grid.drop('stability', axis = 1)
y = electrical_grid['stability']

# Dividir los datos de entrenamiento (train) y datos de prueba (test)
# Especificar que el tamaño de los datos para prueba será el 20% de los registros
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Especificar el tipo de modelo
model = LogisticRegression()

# Entrenar el modelo con los conjuntos de entrenamiento (train)
model.fit(X_train, y_train)

# Estima la probabilidad de que un objeto entre en la segunda clase (1)
probabilities = model.predict_proba(X_test)[:,1]

# Predicción para los datos de pruba
predictions = model.predict(X_test)

# Muestra todas las métricas estudiadas para la predicción resultante
print('Métricas para la predicción automática realizada con predict')
print('Accuracy: {:.2f}'.format(accuracy_score(y_test, predictions)))
print('Precision: {:.2f}'.format(precision_score(y_test, predictions)))
print('Recall: {:.2f}'.format(recall_score(y_test, predictions)))
print('F1: {:.2f}\n'.format(f1_score(y_test, predictions)))

# establece el nuevo umbral
threshold = 0.4

# Estandarizar las probabilidades a 0 o 1.
custom_predictions = [0 if i < threshold else 1 for i in probabilities]

# muestra todas las métricas para la predicción con un nuevo umbral
print('Métricas para la predicción con umbral personalizado')
print('Accuracy for custom: {:.2f}'.format(accuracy_score(y_test, custom_predictions)))
print('Precision for custom: {:.2f}'.format(precision_score(y_test, custom_predictions)))
print('Recall for custom: {:.2f}'.format(recall_score(y_test, custom_predictions)))
print('F1 for custom: {:.2f}'.format(f1_score(y_test, custom_predictions)))
```

¡Es correcto!

¡Funcionó! El recall de la métrica prioritaria mejoró, pero no afectó tanto a accuracy ni a la F1. ¡Tenemos que experimentar con este umbral! En los negocios, ajustar el umbral es fundamental para la toma de decisiones (si el objeto pertenece a la clase "0" o "1").
