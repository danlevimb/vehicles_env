```python
# Ejemplo de estandarización de muestras.
# Importar librerías 

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Ruta del dataset
file_path = "C:/Git/vehicles_env/datasets/dataset_facebook_cosmetics_us.csv"

# Cargar dataframe
fb = pd.read_csv(file_path, sep=';')
```

```python
# divide los datos en características (la matriz X) y una variable objetivo (y)
X = fb.drop('Total Interactions', axis = 1)
y = fb['Total Interactions']

# Imprime los nombres de las características del conjunto de datos
#X.info()

# Dividir los datos de entrenamiento y datos de prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Imprimir la media y la desviación standard de la función 'Page total likes'
print('Media para entrenamiento:', np.mean(X_train.iloc[:,0]))
print('Desviación standard para entrenamiento:', np.std(X_train.iloc[:,0]))

# Crear la instancia del objeto estandarizador
scaler = StandardScaler()

# Entrenar el scaler en el conjunto de entrenamiento usando el método fit
scaler.fit(X_train)

# Estandarizar el conjunto de entrenamiento utilizando el método transform
X_train_st = scaler.transform(X_train)
X_test_st = scaler.transform(X_test)

print(' \nMedia para entrenamiento (estandarizado)', np.mean(X_train_st[:,0]))
print('Desviación standard para entrenamiento (estandarizado)', np.std(X_train_st[:,0]))
print('\nMedia para prueba (estandarizado)', np.mean(X_test_st[:,0]))
print('Desviación standard para prueba (estandarizado)', np.std(X_test_st[:,0]))
```

¡Es correcto!

Antes de que se estandarizaran los datos, la media de 'Page total likes' para el conjunto de entrenamiento era de cientos de miles y ahora es 0. La desviación estándar era de decenas de miles antes de la estandarización y ahora es 1. Sin embargo, para el conjunto de validación, la media después de la estandarización no es exactamente 0 y la desviación estándar no es exactamente 1. Esto se debe a que la distribución de la característica cambió levemente (pero no drásticamente) para el conjunto de validación.
