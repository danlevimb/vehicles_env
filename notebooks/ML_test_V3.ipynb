```python
# Ejemplo de trazado de histogramas para visualizar las diferencias de los datos de entrenamiento / prueba
# Importar librerías 

import pandas as pd
from sklearn.model_selection import train_test_split
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Ruta del dataset
file_path = "C:/Git/vehicles_env/datasets/dataset_facebook_cosmetics_us.csv"

# Cargar dataframe
fb = pd.read_csv(file_path, sep=';')
```

```python
# Crear el dataframe con las características (matriz X)
X = fb.drop(columns='Total Interactions')

# Crear el dataframe con la variable objetivo
y = fb['Total Interactions']
```

```python
# Dividir los "datos de entrenamiento" y "datos de prueba"
# El parámetro test_size = 0.2 está indicandole que el 80% de los datos serán usados para entrenar
# y el 20% para "datos de prueba"
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state=0)

# histograma de la variable objetivo para el entrenamiento
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
sns.histplot(y_train, bins=30, kde=True)
plt.title('Train — Total Interactions')
plt.xlabel('Total Interactions')
plt.ylabel('Frecuencia')

# histograma de la variable objetivo para la prueba
plt.subplot(1,2,2)
sns.histplot(y_test, bins=30, kde=True)
plt.title('Test — Total Interactions')
plt.xlabel('Total Interactions')
plt.ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

Como podemos ver por los resultados, las distribuciones de la variable objetivo para los conjuntos de prueba y entrenamiento son similares. Si hicieses lo mismo para el resto de las características, podrías decir con certeza si el modelo entrenado podría funcionar mejor con los datos de prueba.
